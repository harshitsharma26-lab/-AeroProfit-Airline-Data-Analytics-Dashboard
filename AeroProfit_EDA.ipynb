```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("up-airline_route_profitability.csv")
```

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])
```

```python
df.describe()
```

```python
total_revenue = df["Total_Revenue"].sum()
total_cost = df["Total_Cost"].sum()
total_profit = df["Profit"].sum()

print("Total Revenue:", round(total_revenue, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
```

```python
df["Month"] = df["Flight_Date"].dt.month
df["Month_Name"] = df["Flight_Date"].dt.strftime("%b")
```

```python
monthly = df.groupby(["Month", "Month_Name"])[
    ["Total_Revenue", "Total_Cost", "Profit"]
].sum().reset_index()

monthly = monthly.sort_values("Month")

monthly
```

```python
plt.figure(figsize=(12, 6))

plt.plot(monthly["Month_Name"], monthly["Total_Revenue"], label="Revenue")
plt.plot(monthly["Month_Name"], monthly["Profit"], label="Profit")

plt.title("Monthly Revenue and Profit")
plt.xlabel("Month")
plt.ylabel("Amount")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
route_profit = df.groupby("Destination").agg(
    Total_Flights=("Flight_Number", "count"),
    Total_Passengers=("Passengers", "sum"),
    Total_Revenue=("Total_Revenue", "sum"),
    Total_Cost=("Total_Cost", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin", "mean"),
    Avg_Load_Factor=("Load_Factor", "mean")
).reset_index()

route_profit = route_profit.sort_values(
    "Total_Profit",
    ascending=False
)

route_profit
```

```python
top_routes = route_profit.head(10)

top_routes
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    top_routes["Destination"],
    top_routes["Total_Profit"]
)

plt.title("Top 10 Most Profitable Destinations")
plt.xlabel("Destination")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
bottom_routes = route_profit.tail(10).sort_values(
    "Total_Profit",
    ascending=True
)

bottom_routes
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    bottom_routes["Destination"],
    bottom_routes["Total_Profit"]
)

plt.title("Bottom 10 Least Profitable Destinations")
plt.xlabel("Destination")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
aircraft_summary = df.groupby("Aircraft_Type").agg(
    Total_Flights=("Flight_Number", "count"),
    Total_Passengers=("Passengers", "sum"),
    Total_Revenue=("Total_Revenue", "sum"),
    Total_Cost=("Total_Cost", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Load_Factor=("Load_Factor", "mean"),
    Avg_Profit_Margin=("Profit_Margin", "mean"),
    Avg_Flight_Hours=("Flight_Hours", "mean")
).reset_index()

aircraft_summary = aircraft_summary.sort_values(
    "Total_Profit",
    ascending=False
)

aircraft_summary
```

```python
top_aircraft = aircraft_summary.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_aircraft["Aircraft_Type"],
    top_aircraft["Total_Profit"]
)

plt.title("Aircraft Type vs Total Profit")
plt.xlabel("Aircraft Type")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    aircraft_summary["Aircraft_Type"],
    aircraft_summary["Avg_Load_Factor"] * 100
)

plt.title("Average Load Factor by Aircraft Type")
plt.xlabel("Aircraft Type")
plt.ylabel("Average Load Factor (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    aircraft_summary["Aircraft_Type"],
    aircraft_summary["Avg_Profit_Margin"]
)

plt.title("Average Profit Margin by Aircraft Type")
plt.xlabel("Aircraft Type")
plt.ylabel("Average Profit Margin (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Load_Factor_Category"] = pd.cut(
    df["Load_Factor"],
    bins=[0, 0.60, 0.80, 1.00],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)
```

```python
load_factor_summary = df.groupby(
    "Load_Factor_Category",
    observed=True
).agg(
    Total_Flights=("Flight_Number", "count"),
    Total_Passengers=("Passengers", "sum"),
    Avg_Load_Factor=("Load_Factor", "mean"),
    Total_Revenue=("Total_Revenue", "sum"),
    Total_Cost=("Total_Cost", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin", "mean")
).reset_index()

load_factor_summary
```

```python
plt.figure(figsize=(10, 6))

plt.bar(
    load_factor_summary["Load_Factor_Category"].astype(str),
    load_factor_summary["Total_Profit"]
)

plt.title("Total Profit by Load Factor Category")
plt.xlabel("Load Factor Category")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["Load_Factor"] * 100,
    df["Profit"],
    alpha=0.5
)

plt.title("Load Factor vs Profit")
plt.xlabel("Load Factor (%)")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python
correlation = df["Load_Factor"].corr(df["Profit"])

print("Correlation between Load Factor and Profit:", round(correlation, 3))
```
